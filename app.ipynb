```python
import pandas as pd
import joblib

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)
```

```python
model_choice = "Logistic Regression"

print("Selected model:", model_choice)
```

```python
filename = f"model/{model_choice.lower().replace(' ', '_')}.pkl"

model = joblib.load(filename)
scaler = joblib.load("model/scaler.pkl")

print("Model loaded successfully")
```

```python
test_df = pd.read_csv("test_data.csv")

display(test_df.head())

print("Dataset shape:", test_df.shape)
```

```python
X_test = test_df.drop(columns=["target"])
y_test = test_df["target"]

print("Features:", X_test.shape)
print("Target:", y_test.shape)
```

```python
if model_choice in [
    "Logistic Regression",
    "KNN",
    "Support Vector Machine"
]:
    X_test_processed = scaler.transform(X_test)
else:
    X_test_processed = X_test
```

```python
preds = model.predict(X_test_processed)

probs = model.predict_proba(X_test_processed)[:, 1]

print("First 10 predictions:")
print(preds[:10])
```

```python
metrics = {
    "Accuracy": accuracy_score(y_test, preds),
    "AUC": roc_auc_score(y_test, probs),
    "Precision": precision_score(y_test, preds),
    "Recall": recall_score(y_test, preds),
    "F1": f1_score(y_test, preds),
    "MCC": matthews_corrcoef(y_test, preds)
}

metrics_df = pd.DataFrame(
    list(metrics.items()),
    columns=["Metric", "Score"]
)

display(metrics_df)
```

```python
cm = confusion_matrix(y_test, preds, labels=[0, 1])

cm_df = pd.DataFrame(
    cm,
    index=["Actual Malignant", "Actual Benign"],
    columns=["Predicted Malignant", "Predicted Benign"]
)

display(cm_df)
```

```python
report = classification_report(
    y_test,
    preds,
    target_names=["Malignant", "Benign"],
    output_dict=True
)

display(pd.DataFrame(report).transpose())
```
